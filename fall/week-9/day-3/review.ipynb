```python
import matplotlib.pyplot as plt
import csv

teams = []
wins = []
losses = []

with open("hockey_stats.csv", encoding="utf-8") as file:
    fieldnames = ["name","year","wins","losses","goals"]
    reader = csv.DictReader(file, fieldnames)
        
    next(reader)
    
    for record in reader:
        teams.append(record["name"])
        wins.append(int(record["wins"]))
        losses.append(int(record["losses"]))

# Zip data together and sort by wins in descending order
sorted_data = sorted(zip(teams, wins, losses), key=lambda x: x[1])

# Unzip contents back into lists
n = 5
teams = list(map(lambda x: x[0], sorted_data))[:n]
wins = list(map(lambda x: x[1], sorted_data))[:n]
losses = list(map(lambda x: x[2], sorted_data))[:n]

plt.barh(teams, losses, color="orange", label="Losses")
plt.barh(teams, wins, color="green", label="Wins")

plt.title("Bottom 5 Teams (1990-1991)")
plt.grid(axis="x", alpha=0.4, linestyle="--")
plt.xlabel("Wins/Losses")
plt.ylabel("Teams")
plt.legend(loc="upper right")

plt.show()
```
