#### Basic structure of a matplotlib chart

```python
import matplotlib.pyplot as plt

# Simple example: Creating your first chart
# This shows the basic structure of a matplotlib chart

# Data to visualize
hours_studied = [1, 2, 3, 4, 5]
test_scores = [65, 70, 75, 85, 90]

# Create the plot
plt.plot(hours_studied, test_scores)

# Add labels and title
plt.xlabel('Hours Studied')
plt.ylabel('Test Score')
plt.title('Study Hours vs Test Scores')

# Display the chart
plt.show()
```

#### Plotting a bar chart using hockey data

```python
import csv
import matplotlib.pyplot as plt

# Read the hockey data from CSV
teams = []
wins = []

with open('hockey_stats.csv', 'r') as file:
    reader = csv.DictReader(file)
    for row in reader:
        # Only get 1990 data for simplicity
        if row['year'] == '1990':
            teams.append(row['name'])
            wins.append(int(row['wins']))

# Create a bar chart
plt.bar(teams, wins)

# Add labels and title
plt.xlabel('Team')
plt.ylabel('Wins')
plt.title('NHL Team Wins - 1990 Season')

# Rotate team names so they don't overlap
plt.xticks(rotation=45, ha='right')

# Make the layout fit better
plt.tight_layout()

# Display the chart
plt.show()
```

#### Horizontal bar chart (country popluation)

```python
import csv
import matplotlib.pyplot as plt

# Read country data from CSV
countries = []
populations = []

with open('country_data.csv', encoding="utf-8") as file:
    reader = csv.DictReader(file)
    for row in reader:
        countries.append(row['name'])
        populations.append(int(row['population']))

# Combine the data and sort by population
country_data = list(zip(countries, populations))
# Sort the data by population (most to least)
country_data.sort(key=lambda country: country[1], reverse=True)

# Get only the top 10
top_10 = country_data[:10]

# Separate back into lists
top_countries = [country[0] for country in top_10]
top_populations = [country[1] for country in top_10]

# Create a horizontal bar chart
plt.barh(top_countries, top_populations)

# Add labels and title
plt.xlabel('Population')
plt.ylabel('Country')
plt.title('Top 10 Countries by Population')

# Make the layout fit better
plt.tight_layout()

# Display the chart
plt.show()
```

#### Line Chart (trends over time)

```python
import matplotlib.pyplot as plt
import csv

# Example: Line Chart - Shows trends over time
# Use this when you want to see how something changes

# Read hockey data and track one team's performance over years
bruins_years = []
bruins_wins = []

with open('hockey_stats.csv', encoding="utf-8") as file:
    reader = csv.DictReader(file)
    for row in reader:
        if row['name'] == 'Boston Bruins':
            bruins_years.append(int(row['year']))
            bruins_wins.append(int(row['wins']))

# Create a line chart
plt.plot(bruins_years, bruins_wins, marker='o')

# Add labels and title
plt.xlabel('Year')
plt.ylabel('Wins')
plt.title('Boston Bruins Performance Over Time')

# Add a grid to make it easier to read
plt.grid(True, alpha=0.3)

# Display the chart
plt.show()
```

#### Scatter Plot (relationships)

```python
import csv
import matplotlib.pyplot as plt

# Example: Scatter Plot - Shows relationships between two variables
# Use this when you want to see if two things are connected

wins = []
goals = []
team_names = []

with open('hockey_stats.csv', encoding="utf-8") as file:
    reader = csv.DictReader(file)
    for row in reader:
        if row['year'] == '1990':
            wins.append(int(row['wins']))
            goals.append(int(row['goals']))
            team_names.append(row['name'])

# Create a scatter plot
plt.scatter(goals, wins)

# Add labels and title
plt.xlabel('Goals Scored')
plt.ylabel('Games Won')
plt.title('Do Teams That Score More Goals Win More Games?')

# Add a grid to make it easier to read
plt.grid(True, alpha=0.3)

# Display the chart
plt.show()
```

#### Pie Chart (proportions)

```python
import matplotlib.pyplot as plt
import csv

# Example: Pie Chart - Shows how parts make up a whole
# Use this when you want to show percentages or proportions

# Let's show the population distribution across continents
# For simplicity, we'll categorize by first letter patterns
asia_pop = 0
europe_pop = 0
americas_pop = 0
africa_pop = 0
oceania_pop = 0

# Major countries by region (simplified categorization)
asia_countries = ['China', 'India', 'Indonesia', 'Pakistan', 'Bangladesh', 'Japan', 
                  'Philippines', 'Vietnam', 'Thailand', 'Myanmar [Burma]', 'South Korea',
                  'Afghanistan', 'Iraq', 'Saudi Arabia', 'Malaysia', 'Nepal', 'Yemen']
europe_countries = ['Russia', 'Germany', 'United Kingdom', 'France', 'Italy', 'Spain',
                    'Ukraine', 'Poland', 'Romania']
americas_countries = ['United States', 'Brazil', 'Mexico', 'Colombia', 'Argentina',
                      'Canada', 'Peru', 'Venezuela', 'Chile']
africa_countries = ['Nigeria', 'Ethiopia', 'Egypt', 'Democratic Republic of the Congo',
                    'South Africa', 'Tanzania', 'Kenya', 'Algeria', 'Sudan', 'Uganda',
                    'Morocco', 'Ghana']

with open('country_data.csv', encoding='utf-8') as file:
    reader = csv.DictReader(file)
    for row in reader:
        country = row['name']
        pop = int(row['population'])
        
        if country in asia_countries:
            asia_pop += pop
        elif country in europe_countries:
            europe_pop += pop
        elif country in americas_countries:
            americas_pop += pop
        elif country in africa_countries:
            africa_pop += pop
        else:
            oceania_pop += pop

# Create the pie chart
continents = ['Asia', 'Europe', 'Americas', 'Africa', 'Oceania']
populations = [asia_pop, europe_pop, americas_pop, africa_pop, oceania_pop]

plt.pie(populations, labels=continents, autopct='%1.1f%%')
plt.title('World Population Distribution by Continent (Sample)')

# Display the chart
plt.show()
```
