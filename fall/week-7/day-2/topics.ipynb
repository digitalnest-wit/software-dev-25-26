### Reading lines from a CSV file
If we want to work with the lines of data within a CSV file, we can use `csv.reader()` to create a Reader object. This object can then be iterated over using a loop to access the lines of data.

```python
import csv

with open(file="feed_data.csv") as file:
    # Create a Reader object using the opened file.
    reader = csv.reader(file)

    # Iterate over each row in the reader. Here, I enumerated the reader to 
    # see each row index when printing each row.
    for i, row in enumerate(reader):
        # Do something with the row data... here, we just print each row.
        print(f"{i=} {row=}")
```

### Reading data from a CSV file into a Dictionary
More commonly, we'll want to work with *specific* parts of data instead of each line one at a time. We can achieve this by using `csv.DictReader()` to create a `csv.DictReader` object.

```python
import csv

with open(file="feed_data.csv") as file:
    # Create a DictReader object using the opened file.
    reader = csv.DictReader(file)
    
    for i, item in enumerate(reader):
        print(f"{i=} {item=}")
```

Notice the output of the cell above is a dictionary for each line in the CSV file. `csv.DictReader` creates dictionaries as it reads each line, making it very useful for accessing specific parts of the file.

### Writing a row to a CSV File
We can use `csv.writer()` to create a Writer object for writing data into a file. Then, use `writer.writerow()` to write a row into the opened file.

```python
import csv

# Open the data file in append mode, since we want to add a new row (append).
with open("feed_data.csv", mode="a") as file:
    # Create a Writer object using the opened file.
    writer = csv.writer(file)
    
    # Write a row into the file using the writer.
    row_to_append = ["choco", "4ef67aa31b", "3258", "132", "19", "https://i.pinimg.com/736x/6d/3e/2d/6d3e2db09b92eae597d5e00e329f1f1b.jpg"]
    writer.writerow(row_to_append)
```

### Writing a dictionary to a CSV File
We can use `csv.DictWriter()` to create a `csv.DictWriter` object. This object enables us to easily write dictionaries into the file. Use `writer.writerow()` to write a dictionary into the opened file.

```python
import csv

# Open the data file in append mode
with open("feed_data.csv", mode="a") as file:
    # Create a list with the column names you're going to write (required, order matters!).
    columns = ["author_username", "post_id", "num_views", "num_likes", "num_comments", "post_url"]
    
    # Create a DictWriter object using the opened file. Notice that fieldnames is provided.
    writer = csv.DictWriter(file, fieldnames=columns)
    
    # Write the dictionary to the DictWriter
    writer.writerow({
        "author_username": "choco",
        "post_id": "4ef67aa31b",
        "num_views": "3258",
        "num_likes": "132",
        "num_comments": "19",
        "post_url": "https://i.pinimg.com/736x/6d/3e/2d/6d3e2db09b92eae597d5e00e329f1f1b.jpg",
    })
```
