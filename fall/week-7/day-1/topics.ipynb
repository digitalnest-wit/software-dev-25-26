### Opening a File
Use `open()` to open a file and load it into Python. Set the mode to `"r"` for reading.

> **NOTE**
> 
> Don't forget to close the file after you're done using it. Use `close()` to close a file.

```python
file = "secrets.txt"

f = open(file, mode="r")

# Do something with the file ...

f.close()
```

### Opening a File (using `with` statement)
The `with` statement makes it easier to work with files because it automatically calls the `close()` function for us when needed.

> **NOTE**
> 
> It's highly recommended to use the `with` statement when working with files, as it ensures files are closed even if an error ocurrs.

```python
file = "secrets.txt"

with open(file, mode="r") as f:
    # Do something with the file ...
    s = f.read()

print(s)
```

### Reading from a File
Use `f.read()` to read all the contents of a file into a string.

```python
file = "secrets.txt"

with open(file, mode="r") as f:
    s = f.read()

print(s)
```

Use `f.readlines()` to read all the contents of a file into a list of strings.

```python
file = "secrets.txt"

with open(file, mode="r") as f:
    lines = f.readlines()
    
print(lines)
```

### Writing to a File
To write to a file, first make sure you set the *mode* to `"w"` write when calling `open()`. Then, you can use `f.write()` to write a string into `f`.

> NOTE
>
> In `"w"` mode, if the specified file does not exist, Python will create a new, empty file with that name. If the file does exist, opening it in `"w"` mode will truncate (clear) all its existing content.

```python
file = "secrets.txt"

items = """eggs
ham
bacon
"""

with open(file, mode="w") as f:
    f.write(items)
```

Use `"a"` append mode to append data to the end of a file without truncating (clearing) the existing file contents.

```python
file = "secrets.txt"

items = """apples
bananas
"""

with open(file, mode="a") as f:
    f.write(items)
```

### Checking if a File Exists
Import `path` from the `os` module to use `path.exists()`.

```python
from os import path

file = "shopping-list.txt"

if path.exists(file):
    print(f"The file {file} exists.")
else:
    print(f"The file {file} does not exist!")
```

Often, you'll need to work with file paths. Double check your spelling and `/` forward slashes in your code, and use `path.exists()` to check if files exist.

```python
from os import path

file = "some/path/secrets.txt"
if path.exists(file):
    print(f"The file {file} exists.")
else:
    print(f"The file {file} does not exist!")
```
