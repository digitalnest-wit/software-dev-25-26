#### Conditional Logic

```python
centers = ["salinas", "gilroy", "watsonville", "stockton", "modesto"]

if len(centers) == 5 and centers[-1] == "modesto":
    print("* The length of the list is 5.")
    print("* The last item in the list is 'modesto'.")
else:
    print(f"* The length of the list is {len(centers)}")
    print(f"* The last item in the list is '{centers[-1]}'.")

if "watsonville" in centers:
    print("'watsonville' is an item in the centers list!")
else:
    print("'watsonville' is not an item in the centers list..")
```

#### Lists

```python
centers = ["salinas", "gilroy", "watsonville", "stockton"]

print("All Centers:")
for center in centers:
    print(f"  {center}")

print(f"Length of the list: {len(centers)}")

last_center_in_list = centers[-1]
print(f"Last center in the list: {last_center_in_list}")

# Append a new center to the list:
print("Adding 'modesto'...")
centers.append("modesto")

last_center_in_list = centers[-1]
print(f"Last center in the list, after adding a new center: {last_center_in_list}")

# Get the index of 'gilroy' in the list. If it's not found, -1 will be returned:
index_of_gilroy = centers.index("gilroy")

print(f"'gilroy' is at index {index_of_gilroy}.")
```

#### Dictionaries

```python
members_per_center = {
    "salinas": 23,
    "gilroy": 57,
    "watsonville": 33,
    "stockton": 0,
    "modesto": 25
}

print("Total Members per Center:")
for center, total_members in members_per_center.items():
    print(f"  {center}: {total_members}")

# Access just the centers (the keys in the dict):
print("Just the Centers:")
for center in members_per_center.keys():
    print(f"  {center}")

print("Registering 6 members into 'watsonville'...")
members_per_center["watsonville"] += 6

print("Total Members per Center (printed again):")
for center, total_members in members_per_center.items():
    print(f"  {center}: {total_members}")
```

#### Functions

```python
def must_read_number(prompt: str) -> int:    
    while True:
        try:
            num = int(input(prompt))
            break
        except ValueError:
            print(f"I was expecting an integer.. Try again!")
            continue

    return num

num = must_read_number("Enter a number")
print(f"Got it! You entered '{num}'")
```
